```python

import csv
import io
import json
import zipfile

zip_path = "/content/riyadh-travel-planner-m5-private-ready.zip"

with zipfile.ZipFile(zip_path) as z:
    def read_csv(suffix):
        path = next(
            n for n in z.namelist()
            if n.endswith("/data/" + suffix)
        )
        text = z.read(path).decode("utf-8-sig")
        return list(csv.DictReader(io.StringIO(text)))

    activities = read_csv("activities.csv")
    restaurants = read_csv("restaurants.csv")
    hotels = read_csv("hotels.csv")

    rubric_path = next(
        n for n in z.namelist()
        if n.endswith("/judge_rubric_v1.json")
    )
    prompt_path = next(
        n for n in z.namelist()
        if n.endswith("/judge_prompt_v1.json")
    )

    rubric = json.loads(z.read(rubric_path))
    judge_prompt = json.loads(z.read(prompt_path))

activity_map = {r["place_id"]: r for r in activities}
restaurant_map = {r["place_id"]: r for r in restaurants}
hotel_map = {r["place_id"]: r for r in hotels}

itinerary = record["generation_result"]["itinerary"]

def activity_info(pid):
    r = activity_map[pid]
    return {
        "place_id": pid,
        "name": r["name"],
        "interest_group": r["interest_group"],
        "latitude": r["latitude"],
        "longitude": r["longitude"]
    }

def restaurant_info(pid):
    r = restaurant_map[pid]
    return {
        "place_id": pid,
        "name": r["name"],
        "budget_level": r["budget_level"],
        "latitude": r["lat"],
        "longitude": r["lng"]
    }

def hotel_info(pid):
    r = hotel_map[pid]
    return {
        "place_id": pid,
        "name": r["hotel_name"],
        "budget_level": r["budget_level"],
        "latitude": r["latitude"],
        "longitude": r["longitude"]
    }

evidence = {
    "hotel": hotel_info(itinerary["hotel_id"]),
    "days": [
        {
            "day": day["day"],
            "activities": [
                activity_info(pid)
                for pid in day["activities"]
            ],
            "restaurants": [
                restaurant_info(pid)
                for pid in day["restaurants"]
            ]
        }
        for day in itinerary["days"]
    ]
}

judge_input = {
    "user_input": record["user_input"],
    "itinerary": itinerary,
    "dataset_evidence": evidence
}

print(json.dumps(
    judge_input,
    indent=2,
    ensure_ascii=False
))
```

```python

import json
import io
import zipfile
import time
from google.genai import types

# Read the official saved baseline records
def load_baseline_record(case_id, seed=11):
    with zipfile.ZipFile(zip_path) as outer:
        cohort_name = next(
            n for n in outer.namelist()
            if n.endswith("baseline-v3-20261107.zip")
        )
        cohort_bytes = outer.read(cohort_name)

    with zipfile.ZipFile(io.BytesIO(cohort_bytes)) as cohort:
        record_path = (
            f"baseline-v3-20261107/"
            f"{case_id}_seed{seed}/record.json"
        )
        return json.loads(cohort.read(record_path))


# Build dataset evidence for a saved itinerary
def build_judge_input(saved_record):
    itinerary = saved_record["generation_result"]["itinerary"]

    return {
        "user_input": saved_record["user_input"],
        "itinerary": itinerary,
        "dataset_evidence": {
            "hotel": hotel_info(itinerary["hotel_id"]),
            "days": [
                {
                    "day": day["day"],
                    "activities": [
                        activity_info(pid)
                        for pid in day["activities"]
                    ],
                    "restaurants": [
                        restaurant_info(pid)
                        for pid in day["restaurants"]
                    ]
                }
                for day in itinerary["days"]
            ]
        }
    }


def run_judge(case_id, seed=11):
    saved_record = load_baseline_record(case_id, seed)

    if not saved_record["generation_success"]:
        print(f"{case_id}: Generation failed; skipped")
        return None

    case_input = build_judge_input(saved_record)

    judge_request = (
        judge_prompt["judge_instructions"]
        + "\n\nLOCKED RUBRIC:\n"
        + json.dumps(rubric, ensure_ascii=False)
        + "\n\nCASE TO EVALUATE:\n"
        + json.dumps(case_input, ensure_ascii=False)
    )

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=judge_request,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=1200,
            response_mime_type="application/json"
        )
    )

    result = json.loads(response.text)

    criteria = [
        "personalization",
        "budget_alignment",
        "itinerary_coherence",
        "itinerary_feasibility"
    ]

    for criterion in criteria:
        score = result["scores"][criterion]
        reason = result["reasons"][criterion]

        assert type(score) is int and 1 <= score <= 5
        assert isinstance(reason, str) and reason.strip()

    print(f"\n===== {case_id}_seed{seed} =====")
    print(json.dumps(result, indent=2, ensure_ascii=False))

    return {
        "run_id": saved_record["run_id"],
        "judge_model": "gemini-3.1-flash-lite",
        "judge_result": result
    }


judge_results = []

for case_id in ["G05", "G08"]:
    try:
        result = run_judge(case_id, seed=11)

        if result is not None:
            judge_results.append(result)

    except Exception as e:
        print(f"\n{case_id} ERROR:", type(e).__name__)
        print(str(e)[:800])

    time.sleep(3)
```

```python

import json
import time
import random
import zipfile
import io
from pathlib import Path
from datetime import datetime, timezone
from google.colab import drive
from google.genai import types

# 1. Save results permanently in Google Drive
drive.mount("/content/drive")

output_dir = Path(
    "/content/drive/MyDrive/M5_Judge_Results"
)
output_dir.mkdir(parents=True, exist_ok=True)

results_file = output_dir / "baseline_judge_results.json"
summary_file = output_dir / "baseline_judge_summary.json"

MODEL = "gemini-3.1-flash-lite"

CRITERIA = [
    "personalization",
    "budget_alignment",
    "itinerary_coherence",
    "itinerary_feasibility"
]

# 2. Load existing results to support resuming
if results_file.exists():
    results = json.loads(
        results_file.read_text(encoding="utf-8")
    )
    print("Resuming previous evaluation.")
else:
    results = {}

def save_results():
    temp = results_file.with_suffix(".tmp")
    temp.write_text(
        json.dumps(results, indent=2, ensure_ascii=False),
        encoding="utf-8"
    )
    temp.replace(results_file)

# 3. Load the 60 official baseline records
with zipfile.ZipFile(zip_path) as outer:
    cohort_name = next(
        n for n in outer.namelist()
        if n.endswith("baseline-v3-20261107.zip")
    )
    cohort_bytes = outer.read(cohort_name)

records = []

with zipfile.ZipFile(io.BytesIO(cohort_bytes)) as cohort:
    record_paths = sorted(
        n for n in cohort.namelist()
        if n.startswith("baseline-v3-20261107/")
        and n.endswith("/record.json")
    )

    for path in record_paths:
        records.append(json.loads(cohort.read(path)))

assert len(records) == 60, (
    f"Expected 60 records, found {len(records)}"
)

print("Official baseline records:", len(records))

# 4. Build judge request
def make_request(saved_record):
    case_input = build_judge_input(saved_record)

    return (
        judge_prompt["judge_instructions"]
        + "\n\nLOCKED RUBRIC:\n"
        + json.dumps(rubric, ensure_ascii=False)
        + "\n\nCASE TO EVALUATE:\n"
        + json.dumps(case_input, ensure_ascii=False)
    )

# 5. Validate judge output
def validate_judge_output(output):
    assert isinstance(output, dict)

    for criterion in CRITERIA:
        score = output["scores"][criterion]
        reason = output["reasons"][criterion]

        assert type(score) is int
        assert 1 <= score <= 5
        assert isinstance(reason, str)
        assert reason.strip()

    return True

# 6. Evaluate one saved record
def evaluate_record(saved_record):
    request = make_request(saved_record)

    response = client.models.generate_content(
        model=MODEL,
        contents=request,
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=1600,
            response_mime_type="application/json"
        )
    )

    output = json.loads(response.text)
    validate_judge_output(output)

    scores = output["scores"]
    overall = sum(scores[c] for c in CRITERIA) / 4

    return {
        "run_id": saved_record["run_id"],
        "case_id": saved_record["case_id"],
        "seed": saved_record["seed"],
        "status": "judged",
        "judge_model": MODEL,
        "scores": scores,
        "reasons": output["reasons"],
        "overall_score": round(overall, 3),
        "judged_at_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

# 7. Run with retries and incremental saving
for index, saved_record in enumerate(records, start=1):
    run_id = saved_record["run_id"]

    if run_id in results and results[run_id]["status"] in (
        "judged", "generation_failed"
    ):
        print(f"[{index}/60] {run_id}: Already processed")
        continue

    if not saved_record["generation_success"]:
        results[run_id] = {
            "run_id": run_id,
            "status": "generation_failed",
            "overall_score": None
        }
        save_results()
        print(f"[{index}/60] {run_id}: Generation failed")
        continue

    success = False

    for attempt in range(1, 4):
        try:
            result = evaluate_record(saved_record)
            results[run_id] = result
            save_results()

            print(
                f"[{index}/60] {run_id}: SUCCESS "
                f"({result['overall_score']}/5)"
            )

            success = True
            break

        except Exception as e:
            print(
                f"[{index}/60] {run_id}: "
                f"Attempt {attempt} failed: "
                f"{type(e).__name__}: {str(e)[:180]}"
            )

            if attempt < 3:
                time.sleep(
                    min(60, 5 * (2 ** (attempt - 1)))
                    + random.uniform(0, 2)
                )

    if not success:
        results[run_id] = {
            "run_id": run_id,
            "status": "judge_error",
            "overall_score": None
        }
        save_results()

    time.sleep(3)

# 8. Generate summary
judged = [
    r for r in results.values()
    if r["status"] == "judged"
]

failed_generation = sum(
    r["status"] == "generation_failed"
    for r in results.values()
)

judge_errors = sum(
    r["status"] == "judge_error"
    for r in results.values()
)

summary = {
    "experiment": "M5 Baseline",
    "judge_model": MODEL,
    "total_records": len(records),
    "judged": len(judged),
    "generation_failed": failed_generation,
    "judge_errors": judge_errors,
    "average_overall_score": (
        round(
            sum(r["overall_score"] for r in judged)
            / len(judged), 3
        )
        if judged else None
    ),
    "average_by_criterion": {
        c: (
            round(
                sum(r["scores"][c] for r in judged)
                / len(judged), 3
            )
            if judged else None
        )
        for c in CRITERIA
    }
}

summary_file.write_text(
    json.dumps(summary, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("\n========== FINAL SUMMARY ==========")
print(json.dumps(summary, indent=2))
print("\nSaved in:", output_dir)
```

```python

from google import genai
from google.genai import types
from google.colab import userdata
import json

# Separate client with a longer timeout
retry_client = genai.Client(
    api_key=userdata.get("GEMINI_API_KEY"),
    http_options=types.HttpOptions(timeout=60000)
)

saved_record = next(
    r for r in records
    if r["run_id"] == "G04_seed22"
)

try:
    response = retry_client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=make_request(saved_record),
        config=types.GenerateContentConfig(
            temperature=0,
            max_output_tokens=1600,
            response_mime_type="application/json"
        )
    )

    output = json.loads(response.text)
    validate_judge_output(output)

    overall = sum(
        output["scores"][c] for c in CRITERIA
    ) / 4

    results["G04_seed22"] = {
        "run_id": "G04_seed22",
        "case_id": "G04",
        "seed": 22,
        "status": "judged",
        "judge_model": "gemini-3.1-flash-lite",
        "scores": output["scores"],
        "reasons": output["reasons"],
        "overall_score": round(overall, 3)
    }

    save_results()

    print("SUCCESS:", overall, "/5")
    print(json.dumps(output, indent=2))

except Exception as e:
    print("Still unavailable:", type(e).__name__)
    print(str(e)[:300])
```
